```python
# ==========================================
# PART 1 : Import Library
# ==========================================

import random
import pandas as pd
import sqlite3
from datetime import datetime, timedelta
import matplotlib.pyplot as plt

random.seed(42)


# ==========================================
# CLASS 1 : Member
# ==========================================

class Member:

    def __init__(self, member_id, name, age, package_type, expire_date):
        self.member_id = member_id
        self.name = name
        self.age = age
        self.package_type = package_type
        self.expire_date = expire_date

    def is_package_valid(self):
        today = datetime.now().date()
        return self.expire_date >= today

    def membership_status(self):
        if self.is_package_valid():
            return "Active"
        else:
            return "Expired"


# ==========================================
# CLASS 2 : CheckIn
# ==========================================

class CheckIn:

    def __init__(self, checkin_id, member, checkin_time):
        self.checkin_id = checkin_id
        self.member = member
        self.checkin_time = checkin_time
        self.checkout_time = None

    def check_out(self, checkout_time):
        self.checkout_time = checkout_time

    def calculate_hours(self):

        if self.checkout_time is None:
            return 0

        duration = self.checkout_time - self.checkin_time

        return round(duration.total_seconds() / 3600, 2)


# ==========================================
# CLASS 3 : FitnessGym
# ==========================================

class FitnessGym:

    def __init__(self, gym_name):
        self.gym_name = gym_name
        self.members = []
        self.history = []

    def add_member(self, member):
        self.members.append(member)

    def add_history(self, history):
        self.history.append(history)

    def total_members(self):
        return len(self.members)

    def total_checkin(self):
        return len(self.history)


# ==========================================
# Function 1
# สุ่มชื่อสมาชิก
# ==========================================

def generate_member_name():

    first_names = [
        "Somchai","Anan","Napat","Krit",
        "Suda","Nisa","Ploy","Mint",
        "John","David"
    ]

    last_names = [
        "Smith","Johnson","Sukjai",
        "Saelim","Chan","Brown",
        "Taylor"
    ]

    return f"{random.choice(first_names)} {random.choice(last_names)}"


# ==========================================
# Function 2
# สุ่มประเภทแพ็กเกจ
# ==========================================

def random_package(default_package="Monthly"):

    packages = [
        "Monthly",
        "Quarterly",
        "Yearly"
    ]

    return random.choice(packages)


# ==========================================
# Function 3
# สุ่มวันหมดอายุ
# ==========================================

def random_expire_date(package):

    today = datetime.now().date()

    if package == "Monthly":
        days = 30

    elif package == "Quarterly":
        days = 90

    else:
        days = 365

    return today + timedelta(days=days)


# ==========================================
# Function 4
# สุ่มเวลาเข้าใช้บริการ
# ==========================================

def random_checkin_time():

    today = datetime.now()

    hour = random.randint(6, 20)

    minute = random.randint(0, 59)

    return today.replace(
        hour=hour,
        minute=minute,
        second=0,
        microsecond=0
    )


print("Part 1 Complete")
```

```python
# ==========================================
# PART 2 : Generate Members & Check-in Data
# ==========================================

gym = FitnessGym("FitLife Gym")

TOTAL_MEMBERS = 300

for i in range(1, TOTAL_MEMBERS + 1):

    # สุ่มข้อมูลสมาชิก
    name = generate_member_name()
    age = random.randint(18, 60)

    package = random_package()
    expire_date = random_expire_date(package)

    member = Member(
        member_id=i,
        name=name,
        age=age,
        package_type=package,
        expire_date=expire_date
    )

    gym.add_member(member)

    # สุ่มเวลาเช็กอิน
    checkin_time = random_checkin_time()

    check = CheckIn(
        checkin_id=i,
        member=member,
        checkin_time=checkin_time
    )

    # สุ่มระยะเวลาการใช้งาน 1-4 ชั่วโมง
    hours = random.randint(1, 4)
    minutes = random.randint(0, 59)

    checkout_time = checkin_time + timedelta(
        hours=hours,
        minutes=minutes
    )

    check.check_out(checkout_time)

    gym.add_history(check)

print("================================")
print("Generate Complete")
print("Members :", gym.total_members())
print("Check-ins :", gym.total_checkin())
print("================================")

# แสดงตัวอย่างข้อมูล 5 คนแรก
for history in gym.history:

    print("-------------------------------")
    print("Member ID :", history.member.member_id)
    print("Name :", history.member.name)
    print("Age :", history.member.age)
    print("Package :", history.member.package_type)
    print("Expire :", history.member.expire_date)
    print("Status :", history.member.membership_status())
    print("Check In :", history.checkin_time)
    print("Check Out :", history.checkout_time)
    print("Hours Used :", history.calculate_hours())
```

```python
# ==========================================
# PART 3 : Create DataFrame + CSV + SQLite
# ==========================================

import pandas as pd
import sqlite3

# แปลงข้อมูลจาก Object เป็น Dictionary
records = []

for history in gym.history:

    records.append({

        "checkin_id": history.checkin_id,

        "member_id": history.member.member_id,

        "member_name": history.member.name,

        "age": history.member.age,

        "package_type": history.member.package_type,

        "expire_date": history.member.expire_date,

        "status": history.member.membership_status(),

        "check_in": history.checkin_time,

        "check_out": history.checkout_time,

        "hours_used": history.calculate_hours()

    })

# สร้าง DataFrame
df = pd.DataFrame(records)

print("========== Data Preview ==========")
print(df.head())

# ==========================================
# Save CSV
# ==========================================

csv_file = "fitness_gym_data.csv"

df.to_csv(csv_file, index=False)

print(f"\nCSV saved successfully : {csv_file}")

# ==========================================
# SQLite Database
# ==========================================

db_name = "fitness_gym.db"

conn = sqlite3.connect(db_name)

df.to_sql(
    "checkin_history",
    conn,
    if_exists="replace",
    index=False
)

print(f"SQLite Database saved : {db_name}")

# ทดลองอ่านข้อมูลกลับมา
query = """
SELECT *
FROM checkin_history
LIMIT 5
"""

sample = pd.read_sql_query(query, conn)

print("\n========== Data From SQLite ==========")
print(sample)

conn.close()

print("\nPart 3 Complete")
```

```python
# ==========================================
# PART 4 : Data Analysis with Pandas
# ==========================================

print("=" * 50)
print("PANDAS ANALYSIS")
print("=" * 50)

# ------------------------------------------
# 1. จำนวนสมาชิกแต่ละประเภทแพ็กเกจ
# ------------------------------------------
package_summary = (
    df.groupby("package_type")
      .agg(
          total_members=("member_id", "count"),
          average_hours=("hours_used", "mean")
      )
      .reset_index()
      .sort_values("total_members", ascending=False)
)

print("\n1. Package Summary")
print(package_summary)

# ------------------------------------------
# 2. ชั่วโมงการใช้งานเฉลี่ย
# ------------------------------------------
print("\n2. Average Hours Used")
print(df["hours_used"].mean())

# ------------------------------------------
# 3. สมาชิกที่ใช้งานนานที่สุด
# ------------------------------------------
top_member = (
    df.sort_values("hours_used", ascending=False)
      .head(10)
)

print("\n3. Top 10 Longest Usage")
print(top_member[
    [
        "member_name",
        "package_type",
        "hours_used"
    ]
])

# ------------------------------------------
# 4. จำนวนสมาชิก Active / Expired
# ------------------------------------------
status_summary = (
    df.groupby("status")
      .agg(total=("member_id", "count"))
      .reset_index()
)

print("\n4. Membership Status")
print(status_summary)

# ------------------------------------------
# 5. อายุเฉลี่ยแต่ละแพ็กเกจ
# ------------------------------------------
age_summary = (
    df.groupby("package_type")
      .agg(
          average_age=("age", "mean")
      )
      .reset_index()
)

print("\n5. Average Age by Package")
print(age_summary)

print("\nPart 4 Complete")
```

```python
# ==========================================
# PART 5 : SQL Analysis
# ==========================================

import sqlite3
import pandas as pd

# เชื่อมต่อฐานข้อมูล
conn = sqlite3.connect("fitness_gym.db")

print("=" * 50)
print("SQL ANALYSIS")
print("=" * 50)

# ------------------------------------------
# Query 1 : แสดงสมาชิกทั้งหมด เรียงตามชั่วโมงใช้งาน
# ------------------------------------------
query1 = """
SELECT
    member_name,
    package_type,
    hours_used
FROM checkin_history
ORDER BY hours_used DESC;
"""

result1 = pd.read_sql_query(query1, conn)

print("\nQuery 1")
print(result1.head(10))


# ------------------------------------------
# Query 2 : สมาชิกที่ใช้งานมากกว่า 3 ชั่วโมง
# ------------------------------------------
query2 = """
SELECT
    member_name,
    package_type,
    hours_used
FROM checkin_history
WHERE hours_used > 3;
"""

result2 = pd.read_sql_query(query2, conn)

print("\nQuery 2")
print(result2.head(10))


# ------------------------------------------
# Query 3 : สรุปจำนวนสมาชิกแต่ละแพ็กเกจ
# ------------------------------------------
query3 = """
SELECT
    package_type,
    COUNT(*) AS total_members,
    AVG(hours_used) AS average_hours
FROM checkin_history
GROUP BY package_type
ORDER BY total_members DESC;
"""

result3 = pd.read_sql_query(query3, conn)

print("\nQuery 3")
print(result3)


# ------------------------------------------
# Query 4 : จำนวนสมาชิก Active และ Expired
# ------------------------------------------
query4 = """
SELECT
    status,
    COUNT(*) AS total
FROM checkin_history
GROUP BY status;
"""

result4 = pd.read_sql_query(query4, conn)

print("\nQuery 4")
print(result4)


# ปิดฐานข้อมูล
conn.close()

print("\nPart 5 Complete")
```
