#Part 2 — ออกแบบ Class

จาก Part 1 เรามีข้อมูลหลัก 2 กลุ่ม คือ สมาชิก และ รายการเข้าใช้บริการ

```python
class Member:
    """เก็บข้อมูลสมาชิก Fitness Gym"""

    def __init__(self, member_id, name, is_vip=False):
        self.member_id = member_id
        self.name = name
        self.is_vip = is_vip

    def get_discount_rate(self):
        """คืนค่าอัตราส่วนลดของสมาชิก"""
        if self.is_vip:
            return 0.10
        return 0.0
```

```python
class GymBooking:
    """เก็บข้อมูลรายการเข้าใช้บริการ Fitness Gym"""

    PRICE_PER_HOUR = {
        "Gym": 100,
        "Personal Training": 300,
        "Group Class": 200
    }

    def __init__(self, booking_id, member, hours, service_type):
        self.booking_id = booking_id
        self.member = member
        self.hours = hours
        self.service_type = service_type
        self.status = "รอดำเนินการ"

    def calculate_price(self):
        """คำนวณค่าบริการหลังหักส่วนลด"""

        rate = self.PRICE_PER_HOUR[self.service_type]
        base_price = self.hours * rate
        discount = self.member.get_discount_rate()

        return round(
            base_price * (1 - discount),
            2
        )

    def mark_done(self):
        """เปลี่ยนสถานะเป็นเสร็จแล้ว"""
        self.status = "เสร็จแล้ว"
```

#ถ้าตามตัวอย่างของอาจารย์มันเริ่มจากตรงนี้

```python
import random
import time

random.seed(7)
```

```python
def generate_thai_name():
    """สุ่มชื่อสมาชิก"""

    first_names = [
        "มานี", "สมชาย", "วิชัย",
        "สมหญิง", "ปรีชา", "อรุณี",
        "กิตติ", "นภา", "ธนา", "พลอย"
    ]

    last_names = [
        "ใจดี", "รักเรียน",
        "สายทอง", "ศรีสุข",
        "บุญมี", "แสงทอง",
        "สุขใจ", "เจริญดี"
    ]

    return f"{random.choice(first_names)} {random.choice(last_names)}"
```

```python
def random_hours(min_hours=1, max_hours=8):
    """สุ่มจำนวนชั่วโมงในการใช้บริการ"""

    return round(
        random.uniform(min_hours, max_hours),
        1
    )
```

```python
def random_vip():
    """สุ่มสถานะสมาชิก VIP"""

    return random.choice(["VIP", "ทั่วไป"])
```

```python
def format_currency(amount, symbol="บาท"):
    """จัดรูปแบบตัวเลขเป็นราคา"""

    return f"{amount:,.2f} {symbol}"
```

#Part 4 — สร้างข้อมูล Fitness Gym 300 รายการ

```python
# Part 4: จำลองข้อมูลการเข้าใช้บริการ Fitness Gym 300 รายการ

random.seed(7)

bookings_300 = []

for i in range(1, 301):

    # สร้างสมาชิกทีละคน
    member = Member(
        member_id=i,
        name=generate_thai_name(),
        is_vip=random_vip()
    )

    # สุ่มจำนวนชั่วโมง
    hours = random_hours()

    # สุ่มประเภทบริการ
    service_type = random.choice(
        list(GymBooking.PRICE_PER_HOUR.keys())
    )

    # สร้าง Booking ทีละรายการ
    booking = GymBooking(
        booking_id=i,
        member=member,
        hours=hours,
        service_type=service_type
    )

    # สุ่มสถานะการใช้บริการ
    if random.random() < 0.85:
        booking.mark_done()

    # เก็บ object ลง list
    bookings_300.append(booking)
```

```python
import pandas as pd

rows = []

for booking in bookings_300:
    row = {
        "booking_id": booking.booking_id,
        "member_id": booking.member.member_id,
        "name": booking.member.name,
        "is_vip": booking.member.is_vip,
        "service_type": booking.service_type,
        "hours": booking.hours,
        "price": booking.calculate_price(),
        "status": booking.status
    }

    rows.append(row)

gym_df = pd.DataFrame(rows)

print("จำนวนข้อมูล:", len(gym_df))

gym_df.head()
```

#ตั้งคำถามทางธุรกิจ และวิเคราะห์ด้วย Pandas + SQL

```python
conn = sqlite3.connect("fitness_gym.db")
```

```python
service_count_pandas = (
    gym_df["service_type"]
    .value_counts()
    .reset_index()
)

service_count_pandas.columns = [
    "service_type",
    "total_bookings"
]

service_count_pandas
```

```python
query = """
SELECT
    service_type,
    COUNT(*) AS total_bookings
FROM gym_bookings
GROUP BY service_type
ORDER BY total_bookings DESC
"""

service_count_sql = pd.read_sql_query(
    query,
    conn
)

service_count_sql
```

```python
revenue_pandas = (
    gym_df
    .groupby("service_type")["price"]
    .sum()
    .reset_index()
    .sort_values("price", ascending=False)
)

revenue_pandas.columns = [
    "service_type",
    "total_revenue"
]

revenue_pandas
```

```python
vip_revenue_pandas = (
    gym_df
    .groupby("is_vip")["price"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

vip_revenue_pandas.columns = [
    "is_vip",
    "total_bookings",
    "total_revenue",
    "average_price"
]

vip_revenue_pandas
```

```python
query = """
SELECT
    is_vip,
    COUNT(*) AS total_bookings,
    SUM(price) AS total_revenue,
    AVG(price) AS average_price
FROM gym_bookings
GROUP BY is_vip
ORDER BY total_revenue DESC
"""

vip_revenue_sql = pd.read_sql_query(
    query,
    conn
)

vip_revenue_sql
```

```python
print("บริการที่มีจำนวนการใช้มากที่สุด")
print(service_count_pandas.iloc[0])

print("\nบริการที่สร้างรายได้มากที่สุด")
print(revenue_pandas.iloc[0])

print("\nข้อมูลรายได้ของสมาชิก VIP และสมาชิกทั่วไป")
print(vip_revenue_pandas)
```

```python
# ==============================
# FITNESS GYM MANAGEMENT SYSTEM
# ==============================

import pandas as pd

# เก็บข้อมูลสมาชิก
members = []

# ราคาให้บริการ
service_price = {
    "Gym": 100,
    "Personal Trainer": 300,
    "Group Class": 150
}

# ------------------------------
# เพิ่มสมาชิก
# ------------------------------
def add_member():
    member_id = input("รหัสสมาชิก: ")
    name = input("ชื่อสมาชิก: ")
    status = input("ประเภทสมาชิก (VIP/ทั่วไป): ")

    member = {
        "รหัสสมาชิก": member_id,
        "ชื่อสมาชิก": name,
        "สถานะสมาชิก": status
    }

    members.append(member)
    print("เพิ่มสมาชิกเรียบร้อยแล้ว")


# ------------------------------
# บันทึกการใช้บริการ
# ------------------------------
def use_service():
    member_id = input("รหัสสมาชิก: ")
    service = input(
        "ประเภทบริการ (Gym / Personal Trainer / Group Class): "
    )
    hours = int(input("จำนวนชั่วโมง: "))

    if service in service_price:
        price = service_price[service] * hours

        # ตรวจสอบ VIP
        vip_discount = 0

        for member in members:
            if member["รหัสสมาชิก"] == member_id:
                if member["สถานะสมาชิก"].upper() == "VIP":
                    vip_discount = price * 0.10

        final_price = price - vip_discount

        print("\n----- รายละเอียดการใช้บริการ -----")
        print("ค่าบริการ:", price, "บาท")
        print("ส่วนลด VIP:", vip_discount, "บาท")
        print("ยอดชำระ:", final_price, "บาท")

    else:
        print("ไม่พบบริการนี้")


# ------------------------------
# แสดงสมาชิกทั้งหมด
# ------------------------------
def show_members():
    if len(members) == 0:
        print("ยังไม่มีข้อมูลสมาชิก")
    else:
        df = pd.DataFrame(members)
        display(df)


# ------------------------------
# เมนูหลัก
# ------------------------------
while True:

    print("\n========== FITNESS GYM ==========")
    print("1. สมัครสมาชิก")
    print("2. ใช้บริการ")
    print("3. แสดงข้อมูลสมาชิก")
    print("4. ออกจากระบบ")

    choice = input("เลือกเมนู: ")

    if choice == "1":
        add_member()

    elif choice == "2":
        use_service()

    elif choice == "3":
        show_members()

    elif choice == "4":
        print("ออกจากระบบ")
        break

    else:
        print("กรุณาเลือกเมนู 1-4")
```
