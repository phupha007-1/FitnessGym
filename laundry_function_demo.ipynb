## ส่วนที่ 1 — เตรียม class และฟังก์ชัน (เหมือนตัวอย่างร้านซักรีดเดิม)

```python
import random
import time

random.seed(7)
```

```python
class Customer:
    """ลูกค้าของร้านซักรีด"""

    def __init__(self, customer_id, name, is_vip=False):
        self.customer_id = customer_id
        self.name = name
        self.is_vip = is_vip

    def get_discount_rate(self):
        """คืนค่าส่วนลด (สัดส่วน) ตามสถานะสมาชิก VIP"""
        return 0.10 if self.is_vip else 0.0


class LaundryOrder:
    """ออเดอร์ซักผ้า 1 รายการ"""

    PRICE_PER_KG = {
        "ซัก-อบ": 40,
        "ซัก-อบ-รีด": 60,
        "รีดอย่างเดียว": 30,
    }

    def __init__(self, order_id, customer, weight_kg, service_type):
        self.order_id = order_id
        self.customer = customer
        self.weight_kg = weight_kg
        self.service_type = service_type
        self.status = "รอดำเนินการ"

    def calculate_price(self):
        base_price = self.weight_kg * self.PRICE_PER_KG[self.service_type]
        discount = self.customer.get_discount_rate()
        return round(base_price * (1 - discount), 2)

    def mark_done(self):
        self.status = "เสร็จแล้ว"
```

```python
def generate_thai_name():
    """ฟังก์ชัน: สุ่มชื่อ-นามสกุลลูกค้า -> คืนค่าเป็น string"""
    first_names = ["มานี", "สมชาย", "วิชัย", "สมหญิง", "ปรีชา", "อรุณี"]
    last_names = ["ใจดี", "รักเรียน", "สายทอง", "ศรีสุข"]
    return f"{random.choice(first_names)} {random.choice(last_names)}"


def random_weight(min_kg=1.0, max_kg=8.0):
    """ฟังก์ชัน: สุ่มน้ำหนักผ้า -> คืนค่าเป็น float
    มี default argument: เรียกเฉย ๆ ก็ได้ หรือระบุช่วงเองก็ได้"""
    return round(random.uniform(min_kg, max_kg), 1)


def format_currency(amount, symbol="บาท"):
    """ฟังก์ชัน: จัดรูปแบบตัวเลขเป็นสตริงราคา -> คืนค่าเป็น string"""
    return f"{amount:,.2f} {symbol}"
```

## ส่วนที่ 2 — ทดสอบฟังก์ชันทีละตัว ก่อนเอาไปประกอบเป็นกระบวนการ

ก่อนจะเอาฟังก์ชันไปใช้จริง ควรลองเรียกดูเฉย ๆ ทีละตัวก่อน เพื่อดูว่า **input ที่ใส่เข้าไป**
กับ **output ที่ได้กลับมา** ตรงกับที่ออกแบบไว้หรือไม่ — นี่คือหลักการพื้นฐานของการเขียนฟังก์ชัน:
กำหนด parameter ชัดเจน แล้ว return ค่าที่ใช้ต่อได้


```python
# เรียก generate_thai_name() 3 ครั้ง -> ทุกครั้งได้ชื่อสุ่มไม่ซ้ำแบบ (แสดงว่าฟังก์ชันทำงานทุกครั้งที่เรียก)
for _ in range(3):
    print("ชื่อที่สุ่มได้:", generate_thai_name())
```

```python
# เปรียบเทียบการเรียก random_weight() แบบไม่ระบุค่า (ใช้ default) กับแบบระบุช่วงเอง
print("เรียกแบบ default (ไม่ใส่ argument):", random_weight())
print("เรียกแบบระบุช่วงเอง (2.0 - 5.0 กก.):", random_weight(2.0, 5.0))
print("เรียกแบบระบุแค่ max_kg (keyword argument):", random_weight(max_kg=3.0))
```

```python
# ฟังก์ชัน format_currency() รับตัวเลข -> คืนค่าเป็นสตริงที่จัดรูปแบบแล้ว
raw_price = 1234.5
print("ก่อนจัดรูปแบบ:", raw_price, type(raw_price))
print("หลังจัดรูปแบบ:", format_currency(raw_price), type(format_currency(raw_price)))
```

## ส่วนที่ 3 — ฟังก์ชันอธิบายขั้นตอนคำนวณราคา (ให้เห็นว่าฟังก์ชันเรียกฟังก์ชัน/method อื่นต่อได้)

ฟังก์ชัน `explain_price_calculation()` ด้านล่างไม่ได้คำนวณราคาเอง แต่ **เรียก method ของ object**
(`order.calculate_price()`, `order.customer.get_discount_rate()`) แล้วนำผลลัพธ์มาพิมพ์อธิบายทีละขั้น —
นี่คือตัวอย่างของ "ฟังก์ชันที่ใช้ผลลัพธ์จากฟังก์ชัน/method อื่นต่อ"


```python
def explain_price_calculation(order):
    """รับ object LaundryOrder 1 ตัว -> พิมพ์อธิบายการคำนวณราคาทีละขั้นตอน"""
    rate = order.PRICE_PER_KG[order.service_type]
    base_price = order.weight_kg * rate
    discount_rate = order.customer.get_discount_rate()
    final_price = order.calculate_price()  # เรียก method จริงของ object เพื่อเทียบคำตอบ

    print(f"   ราคาต่อกิโลของบริการ '{order.service_type}' = {rate} บาท/กก.")
    print(f"   ราคาก่อนหักส่วนลด = {order.weight_kg} กก. x {rate} บาท = {format_currency(base_price)}")
    if discount_rate > 0:
        print(f"   ลูกค้าเป็นสมาชิก VIP -> ได้ส่วนลด {discount_rate*100:.0f}%")
    else:
        print("   ลูกค้าไม่ใช่สมาชิก VIP -> ไม่มีส่วนลด")
    print(f"   ราคาสุทธิ = {format_currency(final_price)}")
    return final_price
```

## ส่วนที่ 4 — จำลอง "ลูกค้า 1 คนเดินเข้าร้าน" แบบ step-by-step

ตอนนี้เอาฟังก์ชันทั้งหมดที่มีมาประกอบกันเป็น **1 กระบวนการทางธุรกิจ** ผ่านฟังก์ชันเดียวชื่อ
`simulate_customer_visit()` — ฟังก์ชันนี้เรียกฟังก์ชันอื่น ๆ ที่สร้างไว้ก่อนหน้าเรียงกันเป็นลำดับ
เหมือนพนักงานหน้าร้านทำงานจริง: ทักทาย → ชั่งผ้า → เลือกบริการ → คำนวณราคา → ออกใบเสร็จ


```python
def simulate_customer_visit(order_id, customer, pause=0.0):
    """จำลองขั้นตอนทั้งหมดตอนลูกค้า 1 คนเข้าร้าน แล้วคืนค่า object LaundryOrder ที่สร้างเสร็จแล้ว
    pause = หน่วงเวลาระหว่างขั้นตอน (วินาที) เพื่อให้เห็นจังหวะเวลาตอน demo สด (default ไม่หน่วงเวลา)"""

    print("=" * 60)
    print(f"🚪 ออเดอร์ #{order_id}: ลูกค้า '{customer.name}' เดินเข้ามาที่ร้าน "
          f"({'สมาชิก VIP' if customer.is_vip else 'ลูกค้าทั่วไป'})")
    time.sleep(pause)

    weight = random_weight()  # เรียกฟังก์ชันสุ่มน้ำหนัก
    print(f"⚖️ พนักงานชั่งน้ำหนักผ้า: {weight} กก.")
    time.sleep(pause)

    service = random.choice(list(LaundryOrder.PRICE_PER_KG.keys()))
    print(f"🧾 ลูกค้าเลือกบริการ: '{service}'")
    time.sleep(pause)

    order = LaundryOrder(order_id, customer, weight, service)  # สร้าง object ออเดอร์จริง
    print("📦 ระบบสร้างออเดอร์ในระบบเรียบร้อย (สถานะ:", order.status,")")
    time.sleep(pause)

    print("💰 คำนวณราคา:")
    final_price = explain_price_calculation(order)  # เรียกฟังก์ชันอธิบายราคาที่เขียนไว้ก่อนหน้า
    time.sleep(pause)

    random_time = random.random()  # สุ่มตัวเลข 0.0 - 1.0 เพื่อจำลองเวลาที่พนักงานทำงานเสร็จ
    time_unit = 720  # 1 หน่วยเวลา = 720 นาที (12 ชั่วโมง) -> เพื่อให้เห็นผลลัพธ์ชัดเจน
    print(f"⏰ พนักงานเริ่มดำเนินการซักผ้า... ผ่านไป {random_time*time_unit:.0f} นาที ({random_time*time_unit//60:.0f} ชั่วโมง {(random_time*time_unit)%60:.0f} นาที)")
    if random_time < 0.85:  # จำลองว่าออเดอร์ส่วนใหญ่ทำเสร็จก่อนลูกค้ามารับ
        order.mark_done()
        print(f"✅ งานเสร็จแล้ว! สถานะออเดอร์เปลี่ยนเป็น: '{order.status}'")
    else:
        print(f"⏳ ยังอยู่ระหว่างดำเนินการ สถานะออเดอร์: '{order.status}'")

    print(f"🧾 ใบเสร็จ #{order_id}: ลูกค้า {customer.name} | {weight} กก. | {service} | "
          f"ยอดชำระ {format_currency(final_price)}")

    return order
```

```python
# --- เรียกใช้งานจริงกับลูกค้า 1 คน ---
customer_a = Customer(customer_id=1, name="สมหญิง สายทอง", is_vip=True)
order_a = simulate_customer_visit(order_id=1, customer=customer_a)
```

สังเกตว่าตอนเรียก `simulate_customer_visit()` แค่บรรทัดเดียว แต่ **เบื้องหลังมีฟังก์ชันย่อยหลายตัว
ถูกเรียกต่อกันเป็นลำดับ** (`random_weight`, `random.choice`, `explain_price_calculation`,
`order.calculate_price`, `order.mark_done`) — นี่คือหัวใจของการออกแบบโปรแกรมด้วยฟังก์ชัน:
แตกปัญหาใหญ่เป็นฟังก์ชันย่อย ๆ ที่แต่ละตัวทำหน้าที่เดียวชัดเจน แล้วค่อยประกอบกัน

## ส่วนที่ 5 — จำลองลูกค้าหลายคนเดินเข้าร้านต่อเนื่องกัน

ทีนี้ลองเรียก `simulate_customer_visit()` ซ้ำหลาย ๆ ครั้งผ่าน loop เพื่อดูว่าฟังก์ชันเดียวกัน
**ทำงานซ้ำได้ถูกต้องทุกครั้ง** โดยแต่ละครั้งได้ผลลัพธ์ต่างกันไปตามการสุ่ม — นี่คือหลักการเดียวกับ
ที่ใช้จำลองข้อมูล 300 รายการในโน้ตบุ๊กหลัก เพียงแต่ตรงนี้ทำแค่ไม่กี่คนเพื่อให้เห็นภาพทีละขั้น


```python
walk_in_customers = [
    Customer(customer_id=2, name="มานี ใจดี", is_vip=False),
    Customer(customer_id=3, name="วิชัย ศรีสุข", is_vip=False),
    Customer(customer_id=4, name="ปรีชา แก้วมณี", is_vip=True),
]

completed_orders = []  # เก็บผลลัพธ์ของทุกออเดอร์ที่ simulate ในรอบนี้

for i, cust in enumerate(walk_in_customers, start=2):
    order = simulate_customer_visit(order_id=i, customer=cust)
    completed_orders.append(order)

print("=" * 60)
print(f"🏁 จบรอบสาธิต — วันนี้มีลูกค้าเข้าร้านทั้งหมด {len(completed_orders)} คน (ไม่รวมออเดอร์ #1 ก่อนหน้า)")
```

```python
all_demo_orders = [order_a] + completed_orders
```
